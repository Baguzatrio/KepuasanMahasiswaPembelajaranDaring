```python
def hitung_probkelas (target = dataset['hasil']):
  puas_count = dataset[target.name].value_counts().get('Puas', 0)
  tidakpuas_count = dataset[target.name].value_counts().get('Tidak Puas', 0)
  total_count = len(dataset)

  prob_puas = puas_count / total_count
  prob_tidakpuas = tidakpuas_count / total_count

  return {'Puas': prob_puas, 'Tidak Puas': prob_tidakpuas}

probkelas = hitung_probkelas()
print(probkelas)
```
